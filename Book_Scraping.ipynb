```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
```

```python
driver = webdriver.Chrome()
books = []
ratings = {"One": 1, "Two": 2, "Three": 3, "Four": 4, "Five": 5}
for page in range(1, 6):
    driver.get("https://books.toscrape.com/catalogue/page-" + str(page) + ".html")

    items = driver.find_elements(By.XPATH, "//article[@class='product_pod']")

    for item in items:

        link = item.find_element(By.XPATH, ".//h3/a")
        title = link.get_attribute("title")
        url = link.get_attribute("href")

        price_text = item.find_element(By.XPATH, ".//p[@class='price_color']").text
        price = float(price_text.replace("£", ""))

        rating_class = item.find_element(By.XPATH, ".//p[contains(@class,'star-rating')]").get_attribute("class")
        rating = ratings[rating_class.split()[1]]

        stock = item.find_element(By.XPATH, ".//p[@class='instock availability']").text
    
        books.append({ "title": title,"price": price, "rating": rating, "in_stock": stock, "url": url })

driver.quit()

```

```python
df = pd.DataFrame(books)
df.to_csv("books.csv", index=False)
print("Saved", len(df), "books")
```
